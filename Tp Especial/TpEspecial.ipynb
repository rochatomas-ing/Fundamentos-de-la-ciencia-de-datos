```python
import pandas as pd

df = pd.read_excel("25-testcar-2026-01-21.xlsx")

df
```

```python
print('Cantidad de filas y columnas:', df.shape)
df.head()
df.info()
```

La base contiene 4.287 filas y 67 columnas. Las filas son registros de pruebas y las columnas describen características técnicas del vehículo, información de la prueba, economía de combustible y emisiones.
Se observan identificadores de vehículos y modelos, por lo que conviene distinguir entre cantidad de registros y cantidad de vehículos/modelos únicos.

```python
#Analizamos el porcentaje de nulos que posee cada variable:

porcentaje_nulos = (df.isna().sum() / len(df)) * 100
print("Top 15 columnas con más valores nulos:")
porcentaje_nulos[porcentaje_nulos > 0].sort_values(ascending=False).head(67)
```

Podemos observar que algunas variables poseen un porcentaje alto de valores nulos, los cuales podrian ser descartados. Las primeras 8 variables que aparecen poseen mas de un 90% de valores nulos. Las siguientes tienen entre un 15% y un 66%. Este tipo de variables se pueden analizar para ver si sus datos son relevantes.
