### Analisis de las columnas del dataset:

Vemos que el dataset recopila datos de personas, mas particularmente está enfocado a la salud ya que mide variables que permiten analizar la misma. En la descripcion del dataset, podemos ver que está enfocado a medir la obesidad en mexico, peru y colombia.

AGE: Edad. Variable cuantitativa discreta.

Gender: Genero. Variable cualitativa nominal.

Height: Altura. Variable cuantitativa continua.

Weight: Peso. Variable cuantitativa nominal.

CALC: Consumision de alcohol. Variable cualitativa nominal.

FAVC: Frecuencia en consumir comida altamente calorica(si o no). Variable cualitativa dicotomica.

FCVC: Frecuencia en consumir vegetales(nunca, aveces, siempre). Variable cualitativa nominal.

NCP: Cantidad de comidas principales: Variable cuantitativa discreta.

SCC: Monitoreo del consumo de calorias(si o no). Variable cualitativa dicotomica.

SMOKE: Fuma(si o no). Variable cualitativa dicotomica.

CH20: Cantidad de agua bebida diaria. Variable cualitativa ordinal.

family_history_with_overweight(si o no): Variable cualitativa dicotomica.

FAF: Frecuencia de actividad fisica: Variable cualitativa ordinal.

TUE: Tiempo usando dispositivos tecnologicos: Variable cualitativa ordinal.

CAEC: Consumision entre comidas: Variable cualitativa ordinal.

MTRANS: Uso de transporte: Variable cualitativa nominal.

NObeysedad: Tipo de obesidad: Variable cualitativa ordinal.
