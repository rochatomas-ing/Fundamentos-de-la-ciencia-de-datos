i) Descargar automáticamente ambos archivos

```python
from wget import download
from os import path

if path.exists("netflix_titles.txt"):
    print("El archivo ya existe.")
else:
    download("https://ignaciorlando.github.io/datasets/data-science/netflix_titles.txt")

if path.exists("netflix_titles.csv"):
    print("El archivo ya existe.")
else:
    download('https://ignaciorlando.github.io/datasets/data-science/netflix_titles.csv')
```

ii) Leer el archivo de descripción y mostrar su contenido en pantalla

```python
#Leemos el archivo .txt y lo mostramos en pantalla
with open("netflix_titles.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        print(linea.rstrip())
```
