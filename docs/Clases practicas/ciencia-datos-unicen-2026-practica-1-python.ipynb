## 2 - Variables

Vamos a trabajar algunos ejemplos bien sencillos de manipulación de variables, para que te vayas familiarizando con el tipo de cosas que podemos hacer en el lenguaje.

Las variables se declaran en Python asignándoles valores, y el tipo se determina en función del valor que se le asignan. Si queremos conocer el tipo de dato, podemos usar la función ```type``` (aunque recuerden que es tan pecaminoso como usar un ```instanceof``` en Java!).

Veamos varios ejemplos de declaraciones de variables en Python:

````python
# acá vamos a crear una variable con un entero
variable_1 = 4
print(f"variable_1 = {variable_1} (tipo: {type(variable_1)})")

# acá una con un float
variable_2 = 2.0
print(f"variable_2 = {variable_2} (tipo: {type(variable_2)})")

# acá un string
variable_3 = "esto es un string"
print(f"variable_3 = {variable_3} (tipo: {type(variable_3)})")
````

**Curiosidad! 💡**
Fíjense que formateamos los strings de una forma especial para que uds puedan ver el contenido de las variables. Por defecto, si colocan ```f``` adelante de un string, el lenguaje interpreta que todo lo que encuentre dentro de ese string entre ```{}``` será una variable, y que debe reemplazar ese texto por el valor que tiene adentro esa variable, convertido a string. Otra alternativa es ser un poco más explícitos, usando el método ```format``` de la clase ```str```: ```"variable_1 = {} (tipo: {})".format(variable_1, type(variable_1))"```

Para asignar el valor de una variable a otra, solamente necesito hacer una asignación básica. Algunos detalles importantes:
* **Python no va a chequear compatibilidad de tipos**, porque no es fuertemente tipado. En lugar de hacer eso, va a copiar el valor y el tipo a la nueva variable, pisando todo lo que hubiera ahí antes. Esto es fantástico y a la vez terrible, porque te transfiere a vos toda la responsabilidad de codear como corresponde. [Confiamos en ti, oh, joven Padawan](https://pm1.aminoapps.com/6266/61ed43a2f851687bd2dc741ae52cce29b07fee2c_00.jpg).
* **Las asignaciones son por copia, no por referencia**, esto quiere decir que no estamos asignando punteros, estamos copiando el contenido de los espacios de memoria.

Veamos ejemplos de asignaciones y declaraciones a partir de otras variables

````python
# imprimamos primero los valores de las variables 1 y 2
print(f"[Antes de la asignación] \t variable_1 es = {variable_1} (tipo: {type(variable_1)})")
print(f"[Antes de la asignación] \t variable_2 es = {variable_2} (tipo: {type(variable_2)})")
# acá estamos asignando a la variable 2 el valor de la variable 1
variable_2 = variable_1
print(f"[Después de la asignación] \t variable_1 es = {variable_1} (tipo: {type(variable_1)})")
print(f"[Después de la asignación] \t variable_2 es = {variable_2} (tipo: {type(variable_2)})")
````

Observen que además de alterarse el valor de la variable 2, alteramos también el tipo. Esto es por lo que decíamos antes sobre la copia del valor y el tipo.

Evaluemos ahora qué pasa con la asignación por copia en lugar de por referencia:

````python
# y si actualizamos la variable 1
variable_1 = 5
# vamos a ver un cambio en la variable 1
print(f"variable_1 es = {variable_1} (tipo: {type(variable_1)})")
# pero no en la variable 2, porque las asignaciones son por copia
print(f"variable_2 es = {variable_2} (tipo: {type(variable_2)})")
````

Como ven, la ```variable_2``` no se alteró a pesar de que cambiamos ```variable_1```, lo que efectivamente muestra que la asignación se hizo por copia.

Hay algunos valores especiales de variables que está bueno que conozcan:
* ```None```: el equivalente al null de C/C++, lo solemos usar mucho para declarar variables que no tienen un valor asignado, y que esperamos tengan algún objeto asociado.
* ```True``` y ```False```: los clásicos para manejo de variables booleanas.
* Valores con decimales: si declaran un número con un punto (por ejemplo, 1.0), le están diciendo a Python que necesitan un valor de punto flotante. Si lo declaran sin el punto, se va a inicializar como entero.

````python
A = None
B = 0.0
C = 0.1
D = 5
E = 5.1

print(f"La variable A = {A} - tipo = {type(A)}")
print(f"La variable B = {B} - tipo = {type(B)}")
print(f"La variable C = {C} - tipo = {type(C)}")
print(f"La variable D = {D} - tipo = {type(D)}")
print(f"La variable E = {E} - tipo = {type(E)}")
````

Una última pavadita: en Python podemos hacer asignaciones múltiples, en una sola línea:

### 3.1. Condicionales ```if```

La sintáxis de los ```if``` es bastante sencilla, se usa la sentencia tradicional, con una condición que puede ser simple o compuesta (usando los operadores lógicos comunes: ```and```, ```or``` y ```not```), y en la que podemos usar los típicos comparadores lógicos. Para más data, [chusmeen acá](https://ellibrodepython.com/if-python).

Veamos un ejemplo:

````python
# armamos 3 variables con notas
math_exam = 5.5
english_exam = 5.0
programming_exam = 0.0

# chequeamos si todas están arriba de 7 para ver si promocionó todas
if math_exam >= 7.0 and english_exam >= 7.0 and programming_exam >= 7.0:
  print("Bien ahí! promocionaste todas, capx!")

# si no, chequeamos si alguna está arriba de 7 para ver si promocionó alguna
elif math_exam >= 7.0 or english_exam >= 7.0 or programming_exam >= 7.0:
  print("Bien ahí! promocionaste alguna, capx!")

# si no, no promocionó ninguna
else:
  print("Che, qué garrón, ni una promocionaste!")
````

Veamos algunos detalles importantes. Para empezar, el tabulado. Aunque sabemos que es una buena práctica organizar el código separando ámbitos con tabulaciones, en Python esto es obligatorio: si no usás tabulaciones, el intérprete te va a mostrar un error. Esto es porque, a diferencia de otros lenguajes en los que se usan llaves, paréntesis o corchetes para manejar ámbito,s en Python los ámbitos se manejan con indentaciones.

Fíjense que cuando queremos combinar un ```else``` con otra condición, podemos usar la palabra reservada ```elif```, que es como un ```else if```.

Si queremos usar el "distinto" ($\neq$), en Python tenemos que usar ```!=```. El comparador de igualdad, por otro lado, es el "==":

````python
if math_exam != 10:
  print("Te hacías el genio en matemática pero no te sacaste 10")
elif math_exam == 10:
  print("Alto nerd resultaste!")
````

Otro detalle importante es que las variables que están en ```None``` se consideran como ```False```, con lo cual podemos hacer cosas del tipo:

````python
# inicializamos una variable para guardar notas en None
grades = None

# chequeamos si está en None
if not grades:
  print("Che, no cargaste ni una sola nota")
````

Podemos hacer también unas operaciones en una línea muy tremendas, que están bárbaras porque hacen que el código se lea prácticamente como si fuera declarativo. Pero ojo, las tenés que saber usar bien o te podés mandar alguna cagada:

### 3.2. Bucles ```for```

Los bucles ```for``` en Python son un poco especiales, y explicarte todos sus secretos nos llevaría la vida entera. Te dejamos [acá un lugar del que podés sacar más data](https://ellibrodepython.com/for-python). Están diseñados para recorrer objetos iterables (listas, arreglos, rangos, enumerados, etc.) Por ende, si queremos hacer un recorrido por un cierto número de iteraciones, vamos a necesitar definir un objeto de tipo ```range```, siguiendo siempre esta lógica: ```range(start, stop, step)```, donde ```start``` es el índice por el que queremos empezar, ```stop``` es el último (no se incluye), y ```step``` es el incremento (si no lo ponemos, se asume que es un 1):

````python
# iteramos desde 0 a 5 en incrementos de 1
for i in range(0, 5, 1):
  print(i)
````

Si queremos romper el ```for```, podemos usar tranquilamente un ```break```:

````python
# iteramos de 0 a 5 en incrementos de 1
for i in range(0, 5):
  # imprimimos el índice
  print(i)
  # si el índice es 3, cortamos el bucle
  if i==3:
    print("Corté acá")
    break
````

Y otro ejemplo copado de cosas que podemos hacer en una linea:

````python
# sumamos los elementos de una lista implícita que vamos armando sobre la marcha
print("Suma del 0 al 10 (sin incluir el 10)")
print(sum(i for i in range(10)))

# o sumamos solamente los pares de esa lista
print("Suma de números pares entre 0 y 10 (sin incluir el 10)")
print(sum(i if i % 2 == 0 else 0 for i in range(10)))
````

### 3.3. Bucles ```while```

El ```while``` en Python es bastante parecido a otros lenguajes, con la sola excepción de la sintaxis. Como dijimos antes, acá te vamos a dar algunos ejemplos, pero vas a encontrar [más cosas acá](https://ellibrodepython.com/while-python).

Veamos un ejemplo:

````python
# inicializamos una variable en 5
x = 5
# mientras sea mayor a 0
while x > 0:
  # decrementamos la variable y la imprimimos
  x -=1
  print(x)
````

Podemos por supuesto anidar estos bucles, e incluso traen un ```else```, que se va a ejecutar siempre que el bucle termine "por causas naturales" (que no lo hayamos roto con un ```break``` o no haya habido una excepción o algo por el estilo):

````python
# inicializamos una variable en 5
x = 5
# mientras sea mayor a 0
while x > 0:
  # decrementamos la variable y la imprimimos
  x -=1
  print(x)
else:
  print("Terminé y no me rompí")
````

## 4 - Funciones

Las funciones en Python son muy fáciles de armar, se declaran simplemente usando la palabra reservada ```def```, y poniéndoles un nombre. En este ejemplo, vamos a declarar dos funciones, ```es_par``` y ```es_impar```, que reciben por parámetro una variable ```valor``` y nos devuelven un valor booleano indicando si ```valor``` es par o impar, respectivamente. Si se fijan, una reutiliza la otra, porque ambas están coexistiendo en el mismo ámbito:

````python
def es_par(valor):
  """
  Esta función devuelve un valor booleano indicando de la variable
  valor es par
  """

  # chequeamos si el valor modulo 2 es igual a 0
  if valor % 2 == 0:
    return True
  else:
    return False


def es_impar(valor):
  """
  Esta función devuelve un valor booleano indicando de la variable
  valor es impar
  """
  return not es_par(valor)
````

Al ejecutar el bloque anterior, lo único que hicimos fue declarar las funciones, pero aunque las dos están en memoria, todavía no las usamos. Probémoslas entonces con algunos valores:

````python
print(es_par(8))
print(es_par(13))
print(es_impar(7))
````

Las funciones pueden recibir cuanto parámetro nos guste, e incluso podemos manejar parámetros por defecto. En este ejemplo, les mostramos una función que te saluda por nombre y apellido:

````python
def saludar(nombre, apellido, mensaje_personalizado="Hola!"):
  """
  Función que saluda a una persona por nombre y apellido, con un mensaje personalizado opcional.

  Parámetros:
    nombre (str): El nombre de la persona.
    apellido (str): El apellido de la persona.
    mensaje_personalizado (str, opcional): Un mensaje personalizado para saludar. El valor predeterminado es "Hola!".

  Ejemplo de uso:
    saludar("Juan", "Perez")  # Salida: "Hola Juan Perez!"
    saludar("María", "González", "¡Buen día!")  # Salida: "¡Buen día María González!"
  """
  print(f"{mensaje_personalizado} {nombre} {apellido}")

# ejemplo de llamadas a la función
saludar("Juan", "Perez")
saludar("María", "González", "¿Qué onda?")
````

Tené en cuenta que los parámetros por defecto se definen siempre todos al final, con un valor asociado. No podés mezclar parámetros con y sin valores por defecto, primero se ponen todos los que no tienen valores predefinidos, y después los que si.

Además, las funciones en Python pueden retornar más de un elemento a la vez. Para eso, se listan todas las variables que quieras retornar, separadas por coma, dentro del ```return```. Al llamar a la función, si asignás el valor de retorno a una sola variable, vas a dar con una tupla (podés ver más abajo de qué se trata), que podés iterar con un for. Si no, podés desempaquetarlos usando variables separadas con coma, como una asignación múltiple.

Por ejemplo, acá hacemos una función que nos retorna varias propiedades de un círculo, a la que vamos a usar recuperando una tupla, primero, y desempaquetando sus valores, después.

````python
PI = 3.14159

def calcular_propiedades_circulo(radio):
    """
    Calcula y devuelve múltiples propiedades de un círculo.

    Args:
    radio (float): El radio del círculo.

    Returns:
    tuple: Una tupla conteniendo:
        - area (float): El área del círculo.
        - circunferencia (float): La circunferencia del círculo.
        - diametro (float): El diámetro del círculo.
    """
    area = PI * radio ** 2
    circunferencia = 2 * PI * radio
    diametro = 2 * radio
    return area, circunferencia, diametro

# usando la función con una tupla como salida
resultado = calcular_propiedades_circulo(5)
print(resultado)

# desempaquetando los valores retornados uno a uno
a, c, d = calcular_propiedades_circulo(5)
print(f"Área: {a}, Circunferencia: {c}, Diámetro: {d}")
````

## 5 - Estructuras de datos

Python tiene varias estructuras de datos, que es importante que tengamos en cuenta porque nos van a facilitar la vida cuando analicemos datos (que es el objetivo de la materia, btw).

### 5.1. La lista ```list```

Las listas en Python son el equivalente a un arreglo en otro lenguaje: se puede acceder a sus elementos a través de índices como si fueran un arreglo, aunque no tienen tamaño predefinido (pueden crecer o achicarse dinamicamente).

Las listas las podemos declarar usando corchetes, directamente, poniéndole valores adentro o no.

También Python nos ofrece atajos copados, como poder preguntar si una lista está vacía o no directamente preguntando de esta forma:

````python
# declaramos una lista de notas vacía
grades = [10]

# con esto chequeamos si tiene elementos o no
if grades:
  print("La lista de notas tiene cosas")
else:
  print("La lista de notas está vacía")
````

Las listas no necesariamente tienen que incluir elementos del mismo tipo, como en el ejemplo de acá abajo. Además, contamos con métodos y funciones que nos permiten operar con ellas, por ejemplo:

* ```len()```: nos permite chequear la cantidad de elementos en una lista dada.
* ```lista.append(elemento)```: agrega a la lista ```lista``` el elemento ```elemento```, al final. No devuelve nada, actualiza a ```lista```, así que no necesitamos hacer asignaciones.
* ```lista.extend(lista2)```: agrega todos los elementos de ```lista2``` a ```lista```.
* ```lista.remove(elemento)```: busca ```elemento``` en ```lista``` y lo borra (si no lo encuentra tira una excepción de tipo ```ValueError```).
* ```lista.pop(index=-1)```: saca de ```lista``` el elemento que le indique en ```index``` (si no paso nada, saca el último, como si fuera una pila).

Hay muchos métodos más, [que podés encontrar acá](https://ellibrodepython.com/listas-en-python).

Acá armamos un ejemplo que combina de todo:

````python
# declaramos una lista de notas
grades = [10, 9.5, 4.5, None, 7]
print(grades)

# imprimimos el número de notas total
print(len(grades))

# agregamos un 3 a la lista e imprimimos
grades.append(3)
print(grades)

# imprimimos el primer elemento de la lista
print(grades[0])
# imprimimos el último elemento de la lista
print(grades[-1])
# imprimimos los primeros 3 elementos de la lista
print(grades[0:3])

# usemos esta linea para probar el borrado de cosas
#grades.remove("Casa")
````

Para iterar en listas, podemos hacerlo o bien elemento a elemento, o bien usando índices. Veamos los dos ejemplos:

````python
# iteramos por rango desde 0 al tamaño total de grades
for i in range(len(grades)):
  print(f"Nota {i} = {grades[i]}")
````

````python
# o iteramos elemento a elemento
for grade in grades:
  print(f"Nota = {grade}")
````

Podemos combinar esta forma usando transformaciones a enumerados. Los enumerados (```enumerate```) son iteradores que nos devuelven la posición de un elemento como un índice, y el elemento en sí:

````python
# iteramos usando enumerados
for i, grade in enumerate(grades):
  print(f"Nota {i} = {grade}")
````

También podemos combinar dos listas y recorrerlas simultaneamente, usando ```zip```. Zip en inglés es... "bragueta" (o bueno, no, cierre, o cremallera, jajajaj). La idea acá es que al hacer ```zip``` entre dos listas (que tienen que tener el mismo número de elementos), estamos armando una especie de "cremallera" que podemos recorrer sin problemas:

````python
subjects = ["Inglés", "Programación I", "Cálculo I", "Ciencia de datos", "Teoría de la Información"]
grades = [10, 9.5, 4.5, None, 7]

for subject, grade in zip(subjects, grades):
  print(f"{subject} - Nota = {grade}")
````

### 5.3. El diccionario ```dict```

El diccionario es una estructura muy similar al hashmap con el que trabajábamos en Java, que nos permite almacenar en memoria contenido en pares (clave, valor).
Se trata de colecciones de elementos, en los que a cada uno se le asocia una clave ```key``` y un valor ```value```. La ```key``` actúa como clave o identificador dentro de esa estructura. Pueden pensarlo como si se tratara de un índice de un arreglo, pero que en lugar de ser un número, puede ser cualquier objeto.
Tengan en cuenta que las claves de un diccionario tienen que ser únicas, y no pueden repetirse.

Empecemos viendo algunos ejemplos sobre cómo se crean los diccionarios:

````python
# 1. Inicialización con llaves {}
dict1 = {'nombre': 'Ana', 'edad': 25, 'ciudad': 'Mar Chiquita'}

# 2. Usando el constructor dict()
dict2 = dict(nombre='Carlos', edad=30, ciudad='Barcelona')

# 3. A partir de una lista de tuplas
lista_tuplas = [('fruta', 'manzana'), ('color', 'rojo'), ('precio', 1.5)]
dict3 = dict(lista_tuplas)

# 4. Usando dict.fromkeys()
claves = ['a', 'b', 'c']
dict4 = dict.fromkeys(claves, 0)  # Todos los valores serán 0

# Imprimiendo los resultados
print("dict1:", dict1)
print("dict2:", dict2)
print("dict3:", dict3)
print("dict4:", dict4)
````

También hay algunas otras formas más locas de crear diccionarios, usando las formas pythonicas esas que vimos antes, y que son tan "declarativas":

````python
# 5. Diccionarios "por comprensión"
dict5 = {x: x**2 for x in range(5)}

# Imprimiendo los resultados
print("dict5:", dict5)
````

Para acceder a los elementos que les pusimos dentro, podemos usar los corchetes, como si se tratara de un arreglo, pasando siempre la clave como índice. Tengan en cuenta que si el índice que pasan no existe, el diccionario va a lanzar una excepción del tipo ```KeyError```. Acá les va un ejemplo:

````python
# Los primeros dos accesos deberían correr sin problema, el último no
print(dict1["nombre"])
print(dict3["fruta"])
print(dict3["zapato"])
````

Para modificar elementos, basta con asignarle algo a ese elemento como si fuera una variable común y corriente:

````python
print(dict1)
dict1["ciudad"] = "Mar del Plata"
print(dict1)
````

Y, si el elemento no existe, se va a terminar agregando al diccionario. Esto es porque ños diccionarios son dinámicos, lo que significa que podemos agregar más elementos a medida que vayamos necesitando. Acá va un ejemplo:

````python
print(dict1)
dict1["apellido"] = "Gómez"
print(dict1)
````

Para iterar un diccionario, lo más común es hacerlo siguiendo el típico patrón iterable:

````python
for key in dict1:
  print(f"Clave = {key} - Valor = {dict1[key]}")
````

Si quisiéramos recuperar las claves, podemos usar el método ```.keys()```, que nos va a devolver no una lista, si no un objeto de la clase ```dict_keys```. Si lo queremos como lista, tenemos que castearlo:

````python
print(dict1.keys())
print(list(dict1.keys()))
````

## 6 - Manejo de excepciones

Manejar excepciones es algo que todo buen programador o programadora debe saber hacer, tanto para prevenirlas como para lanzarlas o ejecutar acciones correctivas cuando sea necesario.

Las excepciones están definidas en clases en Python. Esto quiere decir que a cada tipo de excepción le corresponde una clase particular. [Acá](https://docs.python.org/3/library/exceptions.html) pueden encontrar la lista de las excepciones más comunes.

Para capturar una excepción, en lugar de usar ```try / catch``` como en Java, se utiliza ```try / except```:

````python
# inicializamos dos variables de ejemplo
a = 5
b = 0

# todo lo que esté dentro del try puede emitir una excepción, y va a ser controlado
# por el except
try:
    c = a/b

# catcheamos el tipo particular de excepción que nos interesa
except ZeroDivisionError:
  # implementamos acá las acciones correctivas
  print("Estás queriendo dividir por cero, amigo")
````

Se pueden agregar tantos ```except``` como querramos, siempre recordando controlar por la excepción particular que nos interesa. Si, por el contrario, no nos interesa un tipo particular de excepción si no cualquier tipo, podemos usar la clase genérica ```Exception```, y chau picho.

Si, por el contrario, lo que queremos hacer no es capturar una excepción, si no emitirla, podemos usar la sentencia ```raise```. Por ejemplo, en este ejemplo testeamos si en un string hay una palabra prohibida, y lanzamos una excepción de tipo ```ValueError``` si eso ocurre:

````python
def verificar_texto(texto):
  palabras_prohibidas = ["spam", "virus", "malware"]

  for palabra in palabras_prohibidas:
      if palabra in texto.lower():
          raise ValueError(f"Se encontró una palabra prohibida: {palabra}")

  print("El texto es válido.")

# Ejemplos de uso
try:
  verificar_texto("Este es un mensaje normal")
  verificar_texto("Cuidado, este mensaje contiene spam")
except ValueError as e:
  print(f"Error: {e}")
````

Hay bastantes otras cosas copadas para aprender de las expeciones, pero con esto ya tienen lo básico. Si quieren ver más, [puede ir por acá](https://ellibrodepython.com/excepciones-try-except-finally).

## 7 - Clases y objetos

Python ofrece la posibilidad de trabajar con clases y objetos de forma muy similar a Java. Algunos detalles importantes a tener en cuenta:

- El constructor acá se llama ```__init__```, y se llama automáticamente cuando se crea una nueva instancia de una clase. Se define dentro de la clase, y se coloca arriba en la definición.

- El ```__init__``` se usa una sola vez, no podemos tener más de un constructor.

- El primer parámetro del ```__init__``` tiene que ser ```self```, que es el equivalente al ```this``` de Java. Hace referencia al objeto en cuestión.

- Los métodos de instancia (es decir, los que trabajan sobre el objeto, lo modifican o hacen uso de sus atributos) deben incluir ```self``` como parámetro de entrada, también. Y siempre va primero este parámetro. Sí o sí tiene que estar, no está implícito.

- Los atributos de un objeto se pueden acceder desde afuera sin necesidad de escribir métodos específicos para ello. Todo es público, así que guarda con eso.

- Los atributos son también variables, por ende heredan todo lo que sabemos de las variables en Python, incluyendo sus tipos dinámicos.

Acá les dejamos un ejemplo en el que creamos una clase ```Libro```, y generamos varias instancias de diferentes libros.

````python
class Libro:
  """
  Representa un libro en una biblioteca.

  Attributes:
    titulo (str): El título del libro.
    autor (str): El autor del libro.
    año (int): El año de publicación del libro.
    disponible (bool): Indica si el libro está disponible para préstamo.

  Methods:
    prestar(): Marca el libro como no disponible.
    devolver(): Marca el libro como disponible.
    info(): Devuelve una cadena con la información del libro.
  """

  def __init__(self, titulo, autor, año):
    """
    Inicializa una nueva instancia de la clase Libro.

    Args:
        titulo (str): El título del libro.
        autor (str): El autor del libro.
        año (int): El año de publicación del libro.
    """
    self.titulo = titulo
    self.autor = autor
    self.año = año
    self.disponible = True  # Por defecto, el libro está disponible

  def prestar(self):
    """Marca el libro como prestado si está disponible."""
    if self.disponible:
        self.disponible = False
        print(f"El libro '{self.titulo}' ha sido prestado.")
    else:
        print(f"Lo siento, '{self.titulo}' no está disponible.")

  def devolver(self):
    """Marca el libro como devuelto."""
    if not self.disponible:
        self.disponible = True
        print(f"El libro '{self.titulo}' ha sido devuelto.")
    else:
        print(f"Error: '{self.titulo}' ya estaba en la biblioteca.")

  def info(self):
    """
    Devuelve una cadena con la información del libro.

    Returns:
        str: Información formateada del libro.
    """
    estado = "disponible" if self.disponible else "prestado"
    return f"'{self.titulo}' por {self.autor} ({self.año}) - {estado}"


# creamos dos instancias de libros
libro1 = Libro("Don Quijote", "Miguel de Cervantes", 1605)
libro2 = Libro("Cien años de soledad", "Gabriel García Márquez", 1967)

# los usamos
print(libro1.info())  # Muestra la información del libro1
libro1.prestar()      # Presta el libro1
print(libro1.info())  # Muestra la información actualizada del libro1

print(libro2.info())  # Muestra la información del libro2
libro2.prestar()      # Presta el libro2
libro2.prestar()      # Intenta prestar el libro2 nuevamente
libro2.devolver()     # Devuelve el libro2
print(libro2.info())  # Muestra la información actualizada del libro2
````

Python incluye soporte para trabajar con métodos de instancia y métodos de clase. Si repasamos de Objetos, recordemos que los método de instancia operan sobre un objeto específico, tienen acceso a los atributos de la instancia a través de ```self```, y son los métodos más comunes. Por el contrario, los métodos de clase operan sobre la clase misma, no sobre instancias específicas.

Para indicar en Python que un método es de clase, se usa el decorador ```@classmethod```. Además, reciben la clase como primer argumento, con ```cls```, no el objeto.

 Acá les vamos a dejar un ejemplo donde declaramos una nueva clase, ```Biblioteca```, con métodos de instancia y de clase. En este ejemplo declaramos dos bibliotecas, una con dos libros y otra con otros dos, y usamos un atributo de clase para llevar la cuenta del total de libros en bibliotecas.

````python
class Biblioteca:

  total_libros = 0  # Atributo de clase

  def __init__(self, nombre):
    self.nombre = nombre
    self.libros = []

  def agregar_libro(self, libro):  # Método de instancia
    self.libros.append(libro)
    Biblioteca.total_libros += 1

  def listar_libros(self):
    print(f"Libros en {self.nombre}:")
    for libro in self.libros:
        print(f"- {libro.titulo}")

  def buscar_por_autor(self, autor):
    return [libro.titulo for libro in self.libros if libro.autor.lower() == autor.lower()]

  @classmethod
  def desde_lista(cls, nombre, lista_libros):  # Método de clase
    biblioteca = cls(nombre)
    for libro in lista_libros:
      biblioteca.agregar_libro(libro)
    return biblioteca

  @classmethod
  def obtener_total_libros(cls):  # Método de clase
    return cls.total_libros


# Usando método de instancia
biblioteca1 = Biblioteca("Biblioteca Central UNICEN")
biblioteca1.agregar_libro(libro1)
biblioteca1.agregar_libro(libro2)

# Usando método de clase
lista_libros = [Libro("1984", "George Orwell", "1949"), Libro("El Principito", "Antoine de Saint-Exupéry", "1943")]
biblioteca2 = Biblioteca.desde_lista("Biblioteca Rivadavia", lista_libros)

# Obteniendo el total de libros usando un método de clase
print(f"Total de libros en todas las bibliotecas: {Biblioteca.obtener_total_libros()}")
````

Obviamente no hay clases sin herencia, ¿no? En Python es bastante sencillo plantearla. Recordemos que una clase que hereda de otra se trae todos sus atributos y métodos, y que puede añadir atributos y métodos extra. A la clase madre se la refiere con la palabra clave ```super()```.

En este ejemplo agregamos un tipo nuevo de biblioteca, la ```BibliotecaInfantil```, que redefine el método ```agregar_libro()``` controlando si los libros son apropiados:

````python
class BibliotecaInfantil(Biblioteca):
  def __init__(self, nombre, edad_maxima):
    super().__init__(nombre)
    self.edad_maxima = edad_maxima

  def agregar_libro(self, libro):
    if self.es_apropiado(libro):
      super().agregar_libro(libro)
    else:
      print(f"El libro '{libro.titulo}' no es apropiado para esta biblioteca infantil.")

  def es_apropiado(self, libro):
    # Supongamos que los libros publicados antes de 1950 no son apropiados para niños
    return libro.año >= 1950

  def obtener_info(self):
    return f"{self.nombre} (Biblioteca Infantil para niños hasta {self.edad_maxima} años)"


# Uso de las clases
libro1 = Libro("El principito", "Antoine de Saint-Exupéry", 1943)
libro2 = Libro("Harry Potter y la piedra filosofal", "J.K. Rowling", 1997)
libro3 = Libro("Cien años de soledad", "Gabriel García Márquez", 1967)

biblioteca_general = Biblioteca("Biblioteca Central")
biblioteca_general.agregar_libro(libro1)
biblioteca_general.agregar_libro(libro2)
biblioteca_general.agregar_libro(libro3)
biblioteca_general.listar_libros()

print("\n")

biblioteca_infantil = BibliotecaInfantil("Biblioteca de los Pequeños", 12)
biblioteca_infantil.agregar_libro(libro1)  # Este no se agregará por ser muy antiguo
biblioteca_infantil.agregar_libro(libro2)
biblioteca_infantil.listar_libros()

print("\n")

print(biblioteca_infantil.obtener_info())
print("Libros de J.K. Rowling:", biblioteca_infantil.buscar_por_autor("J.K. Rowling"))
````

Python no ofrece soporte para crear interfaces per se, como en Java, pero si nos ofrece la posibilidad de trabajar con clases abstractas. Sin embargo, para hacerlo tenemos que:

1. Importar un paquete particular que ya viene instalado en Python, llamado ```abc```, y traernos de él dos elementos:
  - La clase ```ABC```, de la que deben heredar todas las clases que se consideran abstractas, y
  - ```abstractmethod```, que es el decorador que se usa para indicar que un método es abstracto.
2. Agregar a cada método que es abstracto la palabra clave ```pass``` dentro del cuerpo del método.

En este último ejemplo, modificamos lo que venimos usando previamente, definiendo una clase abstracta ```RecursosBiblioteca```, con los métodos abstractos ```prestar``` y ```devolver```, y dos nuevas clases que heredan de ella, ```Libro``` y ```Revista```, que implementan sus propios métodos para prestar y devolver.

````python
from abc import ABC, abstractmethod

class RecursosBiblioteca(ABC):
  def __init__(self, titulo, autor):
    self.titulo = titulo
    self.autor = autor

  @abstractmethod
  def prestar(self):
    pass

  @abstractmethod
  def devolver(self):
    pass

  def mostrar_info(self):
    return f"{self.titulo} por {self.autor}"


class Libro(RecursosBiblioteca):
  def __init__(self, titulo, autor, isbn):
    super().__init__(titulo, autor)
    self.isbn = isbn
    self.prestado = False

  def prestar(self):
    if not self.prestado:
      self.prestado = True
      return f"Libro '{self.titulo}' prestado."
    return f"El libro '{self.titulo}' ya está prestado."

  def devolver(self):
    if self.prestado:
      self.prestado = False
      return f"Libro '{self.titulo}' devuelto."
    return f"El libro '{self.titulo}' no estaba prestado."


class Revista(RecursosBiblioteca):
  def __init__(self, titulo, autor, numero):
    super().__init__(titulo, autor)
    self.numero = numero
    self.disponible = True

  def prestar(self):
    if self.disponible:
      self.disponible = False
      return f"Revista '{self.titulo}' prestada."
    return f"La revista '{self.titulo}' no está disponible."

  def devolver(self):
    if not self.disponible:
      self.disponible = True
      return f"Revista '{self.titulo}' devuelta."
    return f"La revista '{self.titulo}' ya estaba en la biblioteca."


# Uso de las clases
libro = Libro("1984", "George Orwell", "978-0451524935")
revista = Revista("National Geographic", "Various", "Junio 2023")

print(libro.mostrar_info())
print(libro.prestar())
print(libro.prestar())
print(libro.devolver())

print("\n")

print(revista.mostrar_info())
print(revista.prestar())
print(revista.devolver())

# Esto causaría un error:
# recurso = RecursosBiblioteca("Título", "Autor")  # TypeError: Can't instantiate abstract class
````

Si usamos simplemente ```import <nombre de la libreria>```, vamos a traernos todo el contenido de esa librería. Lo recomendamos usar cuando están seguros de que van a usar muchas funcionalidades pero no saben cuáles.
````python
# importamos Pandas
import pandas

# Crear un DataFrame de ejemplo
data = {
    'Nombre': ['Ana', 'Juan', 'María', 'Carlos'],
    'Edad': [25, 30, 22, 28],
    'Ciudad': ['Tandil', 'Rauch', 'Balcarce', 'Las Flores'],
    'Puntuación': [85, 92, 78, 95]
}

df = pandas.DataFrame(data)

# Mostrar el DataFrame
print("DataFrame original:")
print(df)
print("\n")
````

Como ya se habrán dado cuenta, hacer un import de esta forma implica que cada vez que queremos usar algo del paquete tenemos que escribir todo su nombre completo. Es bastante tedioso eso. Para evitarlo, contamos con los alias. Después de la palabra clave ```as```, podemos ponerle un nombre corto al paquete, así seguimos trabajando con ese:

````python
# importamos Pandas con un alias
import pandas as pd

# Crear un DataFrame de ejemplo
data = {
    'Nombre': ['Ana', 'Juan', 'María', 'Carlos'],
    'Edad': [25, 30, 22, 28],
    'Ciudad': ['Tandil', 'Rauch', 'Balcarce', 'Las Flores'],
    'Puntuación': [85, 92, 78, 95]
}

df = pd.DataFrame(data)

# Mostrar el DataFrame
print("DataFrame original:")
print(df)
print("\n")
````

Cuando no queremos traernos todo el paquete si no determinadas funciones específicas, podemos hacerlo indicando ```from <paquete principal> import <subpaquete que nos interesa>```, pudiendo además ponerle un alias a ese subpaquete, como en este ejemplo:

````python
# Importamos solo el subpaquete pyplot de matplotlib
from matplotlib import pyplot as plt

# Creamos algunos datos de ejemplo
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Creamos un gráfico simple
plt.plot(x, y)

# Añadimos título y etiquetas a los ejes
plt.title('Gráfico de ejemplo')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')

# Mostramos el gráfico
plt.show()

# Imprimimos una confirmación
print("El gráfico se ha generado y mostrado.")
````

Finalmente, también podemos traernos un subpaquete de un subpaquete. El código es muy sencillo:

````python
# Importamos StandardScaler del subpaquete preprocessing de sklearn
from sklearn.preprocessing import StandardScaler

# Importamos numpy para crear nuestros datos de ejemplo
import numpy as np

# Creamos algunos datos de ejemplo
data = np.array([[1., -1., 2.],
                 [2., 0., 0.],
                 [0., 1., -1.]])

# Creamos una instancia de StandardScaler
scaler = StandardScaler()

# Ajustamos el scaler a nuestros datos y los transformamos
scaled_data = scaler.fit_transform(data)

# Mostramos los datos originales y los datos escalados
print("Datos originales:")
print(data)
print("\nDatos escalados:")
print(scaled_data)

# Mostramos la media y la desviación estándar de los datos escalados
print("\nMedia de los datos escalados:")
print(scaled_data.mean(axis=0))
print("\nDesviación estándar de los datos escalados:")
print(scaled_data.std(axis=0))
````
